## Lemmatization
- Lemmatization changes a word to its base dictionary form(root word).
- It uses grammar rules and vocabulary to find correct base form.
- The result is always a real word, unlike stemming.
- Used in making chatbots and language models.

Use stemming when
- Speed is important.
- You don't need perfect word accuracy.
- When working on less memory/CPU device.

Use lemmatization when
- Accuracy matters more than speed.
- Building advanced NLP systems using deep learning.

```python
import pandas as pd
messages = pd.read_csv('sms.csv', index_col=0)
```

```python
messages
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
corpus = []
for i in range(0,len(messages)):
  review = re.sub('^a-zA-Z',' ',messages['text'][i])
  review = review.lower()
  review = review.split()
  review = [lemmatizer.lemmatize(word) for word in review if not word in stopwords.words('english')]
  review = ' '.join(review)
  corpus.append(review)
```

```python
corpus
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features=100)
X = tfidf.fit_transform(corpus).toarray()
```
