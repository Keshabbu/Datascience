Data Encoding --> Process of converting non-numeric data to numeric data

OneHotEncoding Encoding

```python
import pandas as pd
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color' : ['red','blue','green','green','red','blue']
})
```

```python
df
```

```python
# Creating the instance of one hot encoder
encoder = OneHotEncoder()
```

```python
# fit the encoder to dataframe and transform the categorical variable
encoded = encoder.fit_transform(df[['color']])
```

```python
encoded
```

```python
import pandas as pd
encoded_df = pd.DataFrame(encoded.toarray(),columns=encoder.get_feature_names_out())
```

```python
df
```

```python
encoded_df
```

```python
pd.concat([df, encoded_df], axis = 1)
```

Target Guided Ordinal Encoding

```python
df = pd.DataFrame({
    'city' : ['New York','London','Paris','Tokyo','New York','Paris'],
    'price' : [200,150,300,250,150,300]
 })
```

```python
df
```

```python
mean_price = df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded'] = df['city'].map(mean_price)
```

```python
df
```
