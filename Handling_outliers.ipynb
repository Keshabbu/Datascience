Outliers impact the machine learning algorithms which decreases the performance and accuracy of algorithms.

5 Number Summary:
- Minimum Value
- Q1 - 25 percentile
- Q2 - Median or 50 percentile
- Q3 - 75 percentile
- Maximumm Value

```python
import numpy as np
```

```python
numbers = [45, 32, 24, 56, 75, 89, 54, 32, 89, 90, 87, 67, 54, 45, 98, 99, 67, 74, 1000, 1100]
```

```python
np.percentile(numbers, 0)
```

```python
np.percentile(numbers, 25)
```

```python
np.percentile(numbers, 50)
```

```python
np.median(numbers)
```

```python
np.percentile(numbers, 75)
```

```python
np.percentile(numbers, 100)
```

[Lower Fence   -          Upper Fence]

```python
minimum, Q1, median, Q3, maximum = np.quantile(numbers, [0, 0.25, 0.5, 0.75, 1])
```

```python
minimum
```

```python
Q1
```

```python
median
```

```python
Q3
```

```python
maximum
```

```python
iqr = Q3 - Q1
```

```python
iqr
```

```python
lower_fence = Q1 - 1.5 * iqr
higher_fence = Q3 + 1.5 * iqr
```

```python
lower_fence, higher_fence
```

```python
numbers
```

```python
outliers = []

for num in numbers:
  if num < lower_fence or num > higher_fence:
    outliers.append(num)
```

```python
outliers
```
