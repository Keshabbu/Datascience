Standardization(Z-Score)

```python
import numpy as np
mean = np.mean(total_bill)
std = np.std(total_bill)
```

```python
mean
```

```python
std
```

```python
mean - std, mean + std
```

```python
data = []
for i in total_bill:
  z_score = (i - mean) / std
  data.append(z_score)
```

```python
print(data)
```

```python
sns.histplot(total_bill)
```

```python
sns.histplot(data)
```

Normalization(Min-Max Scaler)

```python
import seaborn as sns
```

```python
df = sns.load_dataset('taxis')
```

```python
df.head()
```

```python
df.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
min_max = MinMaxScaler()
```

```python
min_max.fit(df[['distance','fare','tip','total']])
```

```python
min_max.transform(df[['distance','fare','tip','total']])
```

```python
df2 = pd.DataFrame(min_max.transform(df[['distance','fare','tip','total']]), columns=['distance','fare','tip','total'])
```

```python
df2.head()
```

```python
max(df2['distance']), min(df2['distance'])
```
